```python
import pandas as pd

ad = pd.read_csv('actual_duration.csv')
ar = pd.read_csv('appointments_regional.csv')
nc = pd.read_excel('national_categories.xlsx')
print(ad)
```

```python
print(ad.columns)
print(ad.shape)
print(ad.dtypes)
print(ad.isnull().sum())
print('---------------------Descriptive Statistics--------------')
print(ad.describe())
print(ad.info())
```

```python
print(ar.columns)
print(ar.shape)
print(ar.dtypes)
print(ar.isnull().sum())
print('---------------------Descriptive Statistics--------------')
print(ar.describe())
print(ar.info())
```

```python
print(nc.columns)
print(nc.shape)
print(nc.dtypes)
print(nc.isnull().sum())
print('---------------------Descriptive Statistics--------------')
print(nc.describe())
print(nc.info())
```

```python
# How many unique Sub-ICB locations are there?
print("\nCount of unique service settings: ", nc['sub_icb_location_name'].value_counts().count())

#Top 5 locations by appointments
print("\nTop 5 Locations by appointments are:", nc['sub_icb_location_name'].value_counts().sort_values(ascending=False).head(5))

#How many service settings, context types, national categories, appointment statuses?
print("\nCount of unique service settings: ", nc['service_setting'].value_counts().sort_values(ascending=False).count())
print("Count of unique context types: ", nc['context_type'].value_counts().sort_values(ascending=False).count())
print("Count of unique national categories: ", nc['national_category'].value_counts().sort_values(ascending=False).count())
print("Count of unique appointment statuses: ", ar['appointment_status'].value_counts().sort_values(ascending=False).count())
```
