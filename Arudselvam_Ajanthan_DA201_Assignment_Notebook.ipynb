### Analyse tweets from Twitter with hashtags related to healthcare in the UK.

```python
# Libraries and settings needed for analysis
import pandas as pd
import seaborn as sns

# Set figure size.
sns.set(rc={'figure.figsize':(15, 12)})

# Set the plot style as white.
sns.set_style('white')

# Maximum column width to display
pd.options.display.max_colwidth = 200
```

```python
# Load the data set.
tweets = pd.read_csv('tweets.csv')

# View the DataFrame.
print(tweets.head())
```

```python
# Explore the metadata.
print(tweets.info())
```

```python
# Explore the data set.
print(tweets.describe())
```

```python
# Would it be useful to only look at retweeted and favourite tweet messages?
# Explain your answer.
print(tweets['tweet_retweet_count'].value_counts())
print(tweets['tweet_favorite_count'].value_counts())
```

```python
# Create a new DataFrame containing only the text.
tweets_text = tweets[tweets['tweet_full_text'].str.contains('#', na=False)]

# View the DataFrame.
print(tweets_text)
```

```python
# Loop through the messages, and create a list of values containing the # symbol.
tags = []
for y in [x.split(' ') for x in tweets['tweet_full_text'].values]:
    for z in y:
        if '#' in z:
            # Change to lowercase.
            tags.append(z.lower())
```

```python
# Display the first 30 records.
print(pd.Series(tags).value_counts().head(30))
```

```python
# Convert the series to a DataFrame in preparation for visualisation.
data = pd.Series(tags).value_counts().reset_index()

# Rename the columns.
data.columns = ['word', 'count']
```

```python
# Fix the count datatype.
data['count'] = data['count'].astype(int)

# View the result.
print(data.head())
```

```python
# Display records where the count is larger than 10.
print(data[data['count'] > 10])
```

```python
# Create a Seaborn barplot indicating records with a count >10 records.
sns.barplot(x='count', y='word', data=data[data['count'] > 10])
```

```python
# Create the plot.
sns.barplot(x='count', y='word', data=data[data['count'] > 10])
plt.title('Hashtags with Count > 10')
plt.xlabel('Count')
plt.ylabel('Hashtag')

# View the barplot.
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# ── Step 1: Load your data and parse the date column ──────────────────────────
df = pd.read_excel('national_categories.xlsx')          # <-- swap in your file
df['appointment_date'] = pd.to_datetime(df['appointment_date'])

# ── Step 2: Build the full expected date range (every calendar day) ───────────
full_range = pd.date_range(start=df['appointment_date'].min(),
                            end=df['appointment_date'].max(),
                            freq='D')

# ── Step 3: Work out which dates in that range never appear in the data ───────
present_dates = pd.to_datetime(df['appointment_date'].unique())
missing_dates = full_range.difference(present_dates)

print(f"Date range checked : {full_range.min().date()} to {full_range.max().date()}")
print(f"Total days expected: {len(full_range)}")
print(f"Days present       : {len(present_dates)}")
print(f"Days missing        : {len(missing_dates)}")

if len(missing_dates) > 0:
    print("\nMissing dates:")
    for d in missing_dates:
        print(" -", d.date())
else:
    print("\nNo missing dates - the data is continuous across the full range.")

# ── Step 4: Plot present vs. missing dates on a timeline ──────────────────────
fig, ax = plt.subplots(figsize=(14, 2.5))

ax.scatter(present_dates, [1] * len(present_dates),
           color='#1C7293', s=10, label='Date present in data')

if len(missing_dates) > 0:
    ax.scatter(missing_dates, [1] * len(missing_dates),
               color='#D64545', s=60, marker='x', linewidths=2,
               label='Date missing from data')

ax.set_yticks([])
ax.set_xlabel('Date')
ax.set_title('Appointment Data Coverage: Missing vs. Present Dates', fontsize=12)
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
plt.xticks(rotation=45, ha='right')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.7), ncol=2, frameon=False)
plt.tight_layout()
plt.show()
```
